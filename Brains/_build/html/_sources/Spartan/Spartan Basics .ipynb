## 2a. How to run the script 

> To run the script there are some key commands you will need to know:
1. Ls- to show you whats in that folder
2. Cd- to change direction into a new folder 
3. scancel (job id)- to cancel a job that is currently running 
4. cd.. - to change direction by going backwards
5. Sbatch (script)- To queue a script to run 
6. squeue -u (your username)- to view the status of your script in the queue( Note: because the server is used all over hte university it can take up to hours for scripts to run)

## Running the script 

Navigate to Punim2033>neuropaths>neuropaths_fmri_bid 

Create a new folder with the subject ID (eg. sub-001)
Then upload the renamed nifti file from your desktop into the corresponding file 
Next open the scripts folder. Then the script named 3D_cortical.sh
Ensure you have all the correct packages loaded, if not follow the steps in teh script to load them. 
Change the array number to that of the participant ID you are wanting to run. 
Note: It doesnt account for the leading zeros. FOr participant 001 you would put 1.

Once you have changed the array number ctl + S to save it 
In your terminal navigate throught the neuropaths folder to ensure the scipt 3D_cortical.sh is in the correct location. Then run the script.

When the scipt done running the output is in the 3D file. There will be a folder named with the participant ID. If you scroll down futher you will find a file called cortical smoothed (for eg: sub-001_cortical_smoothed.stl)
Download this file to your personal computer.
