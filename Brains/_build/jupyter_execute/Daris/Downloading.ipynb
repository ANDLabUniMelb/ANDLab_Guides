# 1b. Downloading images from DARIS

Click on the non-distortion corrected file. 

Then click Dicom (papaya view)

3 views will be shown, we will be sending all 3 views 

You can play around until you have the clearest of all 3 views


## Tips to get the clearest images

 Handy Shortcuts

1. Use the crosshairs found under view

2. Adjust the saturation so that the structures are more distinguishable 
<img src="../_images/Saturation.png" style="width: 50%;">

3. Make sure you don't include the participants eyes. Especially in the axial and coronal planes

4. Make sure you on one side of the membrane that splits the two hemispheres for the sagittal view




Once you have clear images they can be screen snipped and saved as the participant ID followed by A, B, C for each view.

Images are then ready to be sent off to the participant.
