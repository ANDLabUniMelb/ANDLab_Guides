# 1c. Nifti download

**Download the non distortion corrected MP2Rage file**


* Right click on file

* Select download as archive 

* Select transcode from dicom/series to: scroll down to nifti/series
<img src="../_images/Downloading as nifti.png" style="width: 50%;">


* Then select download

* This will open as a Zip file 

* Open this file and click all the way through the file until you reach these 3 files:
<img src ="../_images/Zip file.png" style="width: 50%;"> 

* The only file you need is the NII file

* Rename this file as the participant ID_Tw1
(for example sub-001_T1w.nii)
** Note: Be sure to inclued the capital letters and spaces in the right places. The script is case sensitive and won't run otherwise

* Save this renamed fiel to a location on your personal laptop. You can delete the rest of the downloaded zip file.
