## Using Bambu

First step is import the Fablab Bambu template. If you have yet downloaded it you can find it here:

>[Fab Lab- Bambu Studio template_v1.3mf](https://1705907971-files.gitbook.io/~/files/v0/b/gitbook-x-prod.appspot.com/o/spaces%2F-LXb8cACCUE4sAj3yO3X%2Fuploads%2FSDrmdeW6G5b5wOUBB128%2FFab%20Lab%20-%20Bambu%20Studio%20Template_v1.3mf?alt=media&token=61fa07a1-a487-46bf-8a6d-a3f1eecb3bc6)



Next you need to import the brain in.
Click file> import> stl
Choose the stl file that was save from spartan. This will then open onto the plate.

**Scaling and slicing**

The brain opened on the plate is at 100% scaled size. We want to print it at 50%. To do this click the scale button (below) and change to 50
<img src="../_images/Scale.png" style="width: 50%;">


Next the brain needs to be orientated so that it can be printed most effeciencty as well as with the greatest structural integrity. Click the auto orientate button shown below 
<img src= "../_images/Orientate.png" style ="width: 50%;">

When uploading to print you will need 2 different files, one of the brain and one of the sliced plate containing the brain. 

**Saving the brain**
Selecy file in the top lefthand corner. Then export adn all objects as one Stl.
Save file in one folcation using the Fablabs naming convention. Eg Sub-001_WhitePLA_v1.stl 

**Saving the plate**
In the top righthand corner click slice plate, The template will automatically generate the infil density, supports and slicing dimnesions. Once the brain is sliced the below screen will pop up. Take note of the volume (you will need to input this when submitting a job)
 <img src = "../images/Volume.png" style="width: 50%">

 Then click export sliced plate. Save the file in teh same locaton as the exported brain and again use the same naming convention (eg: Sub-001_WhitePLA_v1.gcode.3mf)

 ## Submitting a job


 Use this link to submit a job
 >[FabLab job submission](https://apps.powerapps.com/play/e/default-0e5bf3cf-1ff4-46b7-9176-52c538c22a4d/a/ccc5d0b6-5c55-4c2e-8a3d-1579bca670f4?tenantId=0e5bf3cf-1ff4-46b7-9176-52c538c22a4d&hint=113fce85-af0c-4bd1-ae03-0187262054c0&sourcetime=1741308265634)

There are 2 different materials we can use to 3D print our models, PLA(FDM) and resin. We will be using PLA as it is FAR cheeper!!!!

Enter 000000 as the subject code as this is not for a class. 
Where it asks for volume enter the volume that you took note of in Bambu.
Once details are entered submit. 
If you are charging straight to themis enter the string here, otherwise you will be emailed a link to pay via card after the model is printed. 
A link will be emailed to you for you to upload your files to. 

Open emailed link. This will take you to 3Dprintos.
When signing in search University of Melbourne, 2 options will come up *choose FabLab*

CLick upload files, then attach both the brain stl and slcied plate 3mf.gcode files. 
Your page should look somthing like this.
<img src = "../imaged/3Dprinteros.png" style="width: 50%">

You will note that it says $10.39 while on Bambu it says $2.45 follow the prices on Bambu. 

You will then get email updates when the printed has been approved and when it is being printed.
If you entered a Themis string you will just get an email to say it is ready to be collected. 
If you didn't add any string you will recieve an email to a link to pay for the print. Open this link and search for the print ID (found in email) in the search bar. 

Once paid you can go down to the FabLab located on the ground floor of the Glyn Davis building. You will need your student/staff card to enter which will have apporved access once you have completed the online induction. Prints are found in the cubbys on the lefthand side of the space. 
If you have any questions the team there are super helpful and happy to answer an questions.
