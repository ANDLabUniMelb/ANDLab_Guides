# 3a. Printing at Melbourne Univeristy 

There are 2 different places that we can 3D print on campus

**FABLAB (design school)**
At the FabLab we haev to pay for any models printed. Cost is calculated per gram. A 50% scaled brain tends to cost around $2.50. They are super accomodating here and are ahppy to print as many brains as we need. Given that they are also a facility for students during their busier peroids (eg Weeks 9-exams) they will prioritise student work, so be prepared for longer wait times. On average prints here take 2-4 business days


** Telstra creator space**
Located on Swanston street across from Stop 1. Here printing is free but they largely prioritise student work and don't like us printing large amounts. Their turn around time is longer often atleast a week. This is a good option if you are only wanting to print one or two things.

# 3a. Printing 

```
There are 2 places that we can 3D print on campus at the FabLab located in the Glyn Davis building or the Telstra Creator space on Swanston street.
```

**FabLab**
At the Fablab we are required to pay for all our prints. The cost is calculated per gram. A 50% scaled model tends to cost around $2.50. The FabLab are more than happy for us to print as many models as we need but during their busy peroids (weeks 8-exams) they will prioritise studnet work leading to longer wait times. Generally their turn around time is 2-4 business days. 

**Telstra creator space**
Printing at the Telstra creator space is free but they place a large emphasis on student work and don't appreciate us printing large amounts in their space. 
Both require you to complete online and/or in person inductions in order to be able to print in the space. They have a longer wait of over a week. 

In order to be able to print and enter both spaces you need to complete online and/or in person inductions. These are free to complete and only take an hour or so.
