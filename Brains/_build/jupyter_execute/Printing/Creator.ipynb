# 3c. Telstra creator space

To print at the Telstra creator space you will need to ensure you have completed all the relevant training modules. 

**Upload to 3Dprinteros**

Login to 3dprinteros using the University search function. This will take you to the factor authentification system. Use your Uni credentials to login here to the one that doesn't include FabLab. 

Click add files and upload the stl file that you downloaded from spartan. 

Click on the uploaded brain, it will open you to look like this:
<img src = "../images/Creator space.png" style="width: 50%">

Click on the brain then scale button in the top left corner. We want to scale to **50%** by adjusting the scale to (%). Note: Adjusting this will keep the rest of the proportions the same just scaled. This is important 

Next your want to orientate it for optimal printing. Use the Rotate button to orientate. During your 3D printing induction you should be taught how to be orientate. Be sure to press on bed when finished roataing then hit save. 
The camera setting can be used to orientate yourself around the space 
Then slice slice the saved model. You will see it has a smaller file size due to the scaling. 
Choose ultimaker 2+ connect and slicer 5

# Key slicing settings 

Be sure to change the following slicing setting from the prefilled option. 
```{admonition}
**Profile**
Change profile to PLA (0.4mm)- Normal 

**Layer Height**
Change to 0.25mm

**Wall thickness**
Can stay the same at 0.8 mm

**Infil density**
Reduce to 7%
```


 BE SURE TO TICK GENERATE SUPPORT
